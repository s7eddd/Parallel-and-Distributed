```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Sample data from your results
explorers = ["Explorer 0", "Explorer 1", "Explorer 2", "Explorer 3"]
original_time = [0.001318, 0.002018, 0.001272, 0.001260]
improved_time = [0.002269, 0.002033, 0.002047, 0.002131]
original_moves = [1279, 1279, 1279, 1279]
improved_moves = [128, 128, 128, 128]
```

```python
x = np.arange(len(explorers))  # the label locations
width = 0.35  # width of the bars
```

```python
# Plot time comparison
plt.figure(figsize=(10, 5))
plt.bar(x - width/2, original_time, width, label='Original Time')
plt.bar(x + width/2, improved_time, width, label='Improved Time')
plt.ylabel('Time (seconds)')
plt.title('Time Comparison per Explorer')
plt.xticks(x, explorers)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Plot move comparison
plt.figure(figsize=(10, 5))
plt.bar(x - width/2, original_moves, width, label='Original Moves')
plt.bar(x + width/2, improved_moves, width, label='Improved Moves')
plt.ylabel('Number of Moves')
plt.title('Move Count Comparison per Explorer')
plt.xticks(x, explorers)
plt.legend()
plt.grid(True)
plt.show()
```
